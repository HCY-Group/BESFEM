# Plotting Material Properties

This notebook reproduces the main formulas from `MaterialProperties.cpp` and plots them for `c` from `0` to `1.0`.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path


MATERIAL_DIR = Path("../inputs/materials")

# Faraday constant from your Constants::Frd
FRD = 96485.3365

# Electrolyte diffusivity constant from Constants::D0
D0 = 0.00489

# Concentration range: double c from 0 to 1.0
c = np.linspace(0.0, 1.0, 1001)

# Avoid singular endpoint behavior for functions that clamp c internally
c_safe = np.clip(c, 1.0e-8, 1.0 - 1.0e-8)

plt.rcParams.update({
    "figure.figsize": (7, 4.5),
    "axes.grid": True,
    "font.size": 12,
})
```

## Helper functions

```python
def load_vector(filename, expected_size=None):
    # Load one-column material property table.
    path = MATERIAL_DIR / filename
    if not path.exists():
        print(f"Missing file: {path}")
        return None

    arr = np.loadtxt(path)
    arr = np.asarray(arr, dtype=float)

    if expected_size is not None and len(arr) != expected_size:
        print(f"Warning: {filename} has {len(arr)} values, expected {expected_size}.")

    return arr


def table_values(x, ticks, data):
    # Python version of GetTableValues with endpoint clamping and linear interpolation.
    return np.interp(x, ticks, data, left=data[0], right=data[-1])


def plot_property(x, y, title, ylabel, xlabel="c"):
    fig, ax = plt.subplots()
    ax.plot(x, y)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    plt.show()


def plot_many(x, series, title, ylabel, xlabel="c"):
    fig, ax = plt.subplots()
    for label, y in series.items():
        if y is not None:
            ax.plot(x, y, label=label)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.legend()
    plt.show()
```

## NMC functions

```python
def nmc_ocv(c):
    return (1.095 * c**2) - (8.234e-7 * np.exp(14.31 * c)) + (4.692 * np.exp(-0.5389 * c))


def nmc_mu(c):
    return -FRD * nmc_ocv(c)


def nmc_i0(c):
    val = -0.2 * (c - 0.37) - 1.559 - 0.9376 * np.tanh(8.961 * c - 3.195)
    return (10.0**val) * 1.0e-3


def nmc_diff(c):
    return (0.0277 - 0.084 * c + 0.1003 * c**2) * 1.0e-8


def nmc_conductivity(c):
    return 0.01929 + 0.7045 * np.tanh(2.399 * c) - 0.7238 * np.tanh(2.412 * c) - 4.2106e-6
```

```python
plot_property(c, nmc_ocv(c), "NMC OCV vs c", "OCV [V]")
plot_property(c, nmc_mu(c), "NMC chemical potential vs c", "mu")
plot_property(c, nmc_i0(c), "NMC exchange current density vs c", "i0 [A/cm²]")
plot_property(c, nmc_diff(c), "NMC diffusivity vs c", "D [cm²/s]")
plot_property(c, nmc_conductivity(c), "NMC conductivity vs c", "conductivity [S/cm]")
```

```python
lfp_ticks = load_vector("LFP_Chm_Pot_Ticks.txt", expected_size=201)
lfp_chm_pot = load_vector("LFP_Chm_Pot.txt", expected_size=201)


def lfp_table_value(c):
    if lfp_ticks is None or lfp_chm_pot is None:
        return None
    return table_values(c, lfp_ticks, lfp_chm_pot)
    return ((-1 * table_values(c, lfp_ticks, lfp_chm_pot)) + 3.4) * -96485.3365

#   double val = ((-1 * GetTableValues(c, Ticks, chmPot)) + 3.4) * -Constants::Frd;


# def lfp_ocv(c):
#     chp = lfp_chp_value(c)
#     if chp is None:
#         return None
#     return -chp + 3.4


def lfp_mu(c):
    chp = lfp_table_value(c)
    if chp is None:
        return None
    return ((-1 * table_values(c, lfp_ticks, lfp_chm_pot)) + 3.4) * -96485.3365

    # return chp * 10

def lfp_ocv(c):
    chp = lfp_table_value(c)
    if chp is None:
        return None
    return -chp + 3.4

def lfp_dmu_dc(c):
    # Same finite-difference spacing as your C++ code
    h = 0.005
    c1 = np.maximum(0.0, c - h)
    c2 = np.minimum(1.0, c + h)
    return (lfp_mu(c2) - lfp_mu(c1)) / (c2 - c1)


def lfp_diff(c):
    return np.full_like(c, 5.0e-14, dtype=float)


# def lfp_mob(c):
#     c_clamped = np.clip(c, 1.0e-8, 1.0 - 1.0e-8)
#     D = lfp_diff(c_clamped)
#     dmu = lfp_dmu_dc(c_clamped)
#     return D / np.abs(dmu)

def lfp_mob(c):
    c_clamped = np.clip(c, 1.0e-8, 1.0 - 1.0e-8)
    D = lfp_diff(c_clamped)
    dmu = lfp_dmu_dc(c_clamped)

    M = D / np.abs(dmu)

    return np.minimum(M, 5.85e-13)


def lfp_i0(c):
    c_clamped = np.clip(c, 1.0e-8, 1.0 - 1.0e-8)
    i0_mA = 2.75 * (1.0 - np.exp(-18.0 * c_clamped)) * (1.0 - np.exp(-18.0 * (1.0 - c_clamped)))
    return 4.0 * i0_mA * 1.0e-6


def lfp_conductivity(c):
    return np.full_like(c, 1.0e-4, dtype=float)

def lfp_mob_smooth(c):

    Mmin = 8.5e-13

    B = 5.0e-14

    A1 = 4.5e-12
    A2 = 4.0e-12

    c1 = 0.06
    c2 = 0.86

    w1 = 0.04
    w2 = 0.04

    return (
        Mmin
        + B*(c - 0.5)**2
        + A1/(1.0 + ((c-c1)/w1)**2)
        + A2/(1.0 + ((c-c2)/w2)**2)
    )

# def lfp_mob_smooth(c):
#     M0 = 8.0e-13      # baseline mobility
#     A  = 5.2e-12      # peak height above baseline
#     c1 = 0.07         # left peak location
#     c2 = 0.86         # right peak location
#     sigma = 0.01     # peak width; larger = smoother/wider

#     return (
#         M0
#         + A * np.exp(-((c - c1)**2) / (2 * sigma**2))
#         + A * np.exp(-((c - c2)**2) / (2 * sigma**2))
#     )
```

## Graphite functions


```python
graphite_ticks = load_vector("C_Li_X_101.txt", expected_size=101)
graphite_ocv_table = load_vector("C_Li_O3_101.txt", expected_size=101)
graphite_i0_table = load_vector("C_Li_J2_101.txt", expected_size=101)
graphite_mob_table = load_vector("C_Li_Mb5_101.txt", expected_size=101)
graphite_mu_table = load_vector("C_Li_M6_101.txt", expected_size=101)


def graphite_ocv(c):
    if graphite_ticks is None or graphite_ocv_table is None:
        return None
    return table_values(c, graphite_ticks, graphite_ocv_table)


def graphite_i0(c):
    if graphite_ticks is None or graphite_i0_table is None:
        return None
    return table_values(c, graphite_ticks, graphite_i0_table) * 1.0e-3


def graphite_mob(c):
    if graphite_ticks is None or graphite_mob_table is None:
        return None
    # Same scaling as your C++ code: Mob(i) *= 100.0 * 2.0/3.0
    return table_values(c, graphite_ticks, graphite_mob_table * 100.0 * 2.0 / 3.0)


def graphite_mu(c):
    if graphite_ticks is None or graphite_mu_table is None:
        return None
    return table_values(c, graphite_ticks, graphite_mu_table -0.15)


def graphite_conductivity(c):
    return np.full_like(c, 3.3, dtype=float)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator, ScalarFormatter


# ============================================================
# LOAD GRAPHITE PROPERTY TABLES
# ============================================================

graphite_ticks = load_vector("C_Li_X_101.txt", expected_size=101)
graphite_ocv_table = load_vector("C_Li_O3_101.txt", expected_size=101)
graphite_i0_table = load_vector("C_Li_J2_101.txt", expected_size=101)
graphite_mob_table = load_vector("C_Li_Mb5_101.txt", expected_size=101)
graphite_mu_table = load_vector("C_Li_M6_101.txt", expected_size=101)


# ============================================================
# GRAPHITE MATERIAL PROPERTY FUNCTIONS
# ============================================================

def graphite_ocv(c):
    if graphite_ticks is None or graphite_ocv_table is None:
        return None

    return table_values(
        c,
        graphite_ticks,
        graphite_ocv_table,
    )


def graphite_i0(c):
    if graphite_ticks is None or graphite_i0_table is None:
        return None

    # Convert from the table units to A/cm^2
    return table_values(
        c,
        graphite_ticks,
        graphite_i0_table,
    ) * 1.0e-3


def graphite_mob(c):
    if graphite_ticks is None or graphite_mob_table is None:
        return None

    # Same scaling used in the C++ implementation:
    # Mob(i) *= 100.0 * 2.0 / 3.0
    scaled_mobility = graphite_mob_table * 100.0 * 2.0 / 3.0

    return table_values(
        c,
        graphite_ticks,
        scaled_mobility,
    )


def graphite_mu(c):
    if graphite_ticks is None or graphite_mu_table is None:
        return None

    return table_values(
        c,
        graphite_ticks,
        graphite_mu_table,
    )


def graphite_conductivity(c):
    return np.full_like(c, 3.3, dtype=float)


# ============================================================
# SCIENTIFIC MULTI-PANEL PLOT
# ============================================================

def plot_graphite_material_properties(
    save_path="graphite_material_properties.pdf",
    dpi=600,
):
    """
    Plot graphite OCV, exchange current density, mobility, and
    chemical potential in a publication-quality 2x2 figure.
    """

    required_tables = {
        "graphite_ticks": graphite_ticks,
        "graphite_ocv_table": graphite_ocv_table,
        "graphite_i0_table": graphite_i0_table,
        "graphite_mob_table": graphite_mob_table,
        "graphite_mu_table": graphite_mu_table,
    }

    missing_tables = [
        name for name, values in required_tables.items()
        if values is None
    ]

    if missing_tables:
        print(
            "Skipping graphite property plot because the following "
            f"tables were not found: {', '.join(missing_tables)}"
        )
        return None, None

    # Use a smooth concentration vector for interpolation.
    c = np.linspace(
        np.min(graphite_ticks),
        np.max(graphite_ticks),
        1000,
    )

    ocv = graphite_ocv(c)
    i0 = graphite_i0(c)
    mobility = graphite_mob(c)
    chemical_potential = graphite_mu(c)

    # --------------------------------------------------------
    # Plot style
    # --------------------------------------------------------

    plt.rcParams.update({
        "font.family": "serif",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 11,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 9,
        "axes.linewidth": 0.8,
        "lines.linewidth": 1.7,
        "savefig.bbox": "tight",
    })

    fig, axes = plt.subplots(
        nrows=2,
        ncols=2,
        figsize=(7.2, 5.8),
        sharex=True,
        constrained_layout=True,
    )

    ax_ocv = axes[0, 0]
    ax_i0 = axes[0, 1]
    ax_mob = axes[1, 0]
    ax_mu = axes[1, 1]

    # --------------------------------------------------------
    # (a) Open-circuit voltage
    # --------------------------------------------------------

    ax_ocv.plot(c, ocv, color="black")
    ax_ocv.set_ylabel(r"Open-circuit voltage, $U_{\mathrm{eq}}$ [V]")
    ax_ocv.set_title("(a) Equilibrium potential", loc="left")

    # --------------------------------------------------------
    # (b) Exchange current density
    # --------------------------------------------------------

    ax_i0.plot(c, i0, color="black")
    ax_i0.set_ylabel(
        r"Exchange current density, $i_0$"
    )
    ax_i0.set_title("(b) Exchange current density", loc="left")

    # --------------------------------------------------------
    # (c) Mobility
    # --------------------------------------------------------

    ax_mob.plot(c, mobility, color="black")
    ax_mob.set_xlabel(r"Lithium site fraction, $X$")
    ax_mob.set_ylabel(r"Mobility, $M$")
    ax_mob.set_title("(c) Lithium mobility", loc="left")

    # --------------------------------------------------------
    # (d) Chemical potential
    # --------------------------------------------------------

    ax_mu.plot(c, chemical_potential, color="black")
    ax_mu.set_xlabel(r"Lithium site fraction, $X$")
    ax_mu.set_ylabel(r"Chemical potential, $\mu$")
    ax_mu.set_title("(d) Chemical potential", loc="left")

    # --------------------------------------------------------
    # Common axis formatting
    # --------------------------------------------------------

    for ax in axes.flat:
        ax.set_xlim(
            np.min(graphite_ticks),
            np.max(graphite_ticks),
        )

        ax.tick_params(
            axis="both",
            which="major",
            direction="in",
            top=True,
            right=True,
            length=5,
            width=0.8,
        )

        ax.tick_params(
            axis="both",
            which="minor",
            direction="in",
            top=True,
            right=True,
            length=2.5,
            width=0.6,
        )

        ax.xaxis.set_minor_locator(AutoMinorLocator())
        ax.yaxis.set_minor_locator(AutoMinorLocator())

        ax.grid(
            True,
            which="major",
            linewidth=0.4,
            alpha=0.25,
        )

    # Use scientific notation where values are very small or large.
    for ax in (ax_i0, ax_mob):
        formatter = ScalarFormatter(useMathText=True)
        formatter.set_powerlimits((-2, 2))
        ax.yaxis.set_major_formatter(formatter)
        ax.ticklabel_format(
            axis="y",
            style="sci",
            scilimits=(-2, 2),
        )

    fig.supxlabel(
        r"Lithium concentration, $X$",
        fontsize=11,
    )

    # Avoid duplicating the shared x-label on the lower panels.
    ax_mob.set_xlabel("")
    ax_mu.set_xlabel("")

    

    # --------------------------------------------------------
    # Save and display
    # --------------------------------------------------------

    fig.savefig(
        save_path,
        dpi=dpi,
        bbox_inches="tight",
    )

    # Also save a high-resolution PNG for Word/PowerPoint previews.
    if save_path.lower().endswith(".pdf"):
        png_path = save_path[:-4] + ".png"
    else:
        png_path = save_path + ".png"

    fig.savefig(
        png_path,
        dpi=dpi,
        bbox_inches="tight",
    )

    plt.show()

    print(f"Saved vector figure to: {save_path}")
    print(f"Saved raster figure to: {png_path}")

    return fig, axes


# ============================================================
# CREATE FIGURE
# ============================================================

fig, axes = plot_graphite_material_properties(
    save_path="graphite_material_properties.pdf",
    dpi=600,
)
```

## carbon functions

The carbon functions below reproduce the current definitions in `MaterialProperties.cpp`:

- equilibrium potential: fitted analytical expression,
- chemical potential: $\mu_{\mathrm{Si}}=-U_{\mathrm{Si}}$,
- exchange-current density: the graphite table is reused,
- diffusivity: constant $1.0\times10^{-10}\ \mathrm{cm^2\,s^{-1}}$,
- conductivity: constant $1.0\ \mathrm{S\,cm^{-1}}$,
- site density: $0.0227$.

A carbon mobility is not currently defined in the C++ material-property switch, so it is not plotted.


```python
# ============================================================
# carbon MATERIAL PROPERTY FUNCTIONS
# ============================================================

def carbon_ocv(c):
    """Current carbon_OCV implementation from MaterialProperties.cpp."""
    c_array = np.asarray(c, dtype=float)
    c_clamped = np.clip(c_array, 0.0, 1.0)

    return (
        1.12165244
        * (1.0 - c_clamped)**1.28022415
        * np.exp(-1.76785792 * c_clamped)
    )


def carbon_mu(c):
    """Current C++ definition: carbon_mu(c) = -carbon_OCV(c)."""
    return -carbon_ocv(c)


def carbon_i0(c):
    """
    carbon currently reuses Graphite_i0 in ExchangeCurrentDensity().
    """
    return graphite_i0(c)


def carbon_diff(c):
    """Constant carbon diffusivity used in MaterialProperties.cpp."""
    c_array = np.asarray(c, dtype=float)
    return np.full_like(c_array, 1.0e-10, dtype=float)


def carbon_conductivity(c):
    """Constant carbon electronic conductivity."""
    c_array = np.asarray(c, dtype=float)
    return np.full_like(c_array, 1.0, dtype=float)


carbon_SITE_DENSITY = 0.0227
```

```python
# ============================================================
# PUBLICATION-QUALITY carbon PROPERTY FIGURE
# ============================================================

def plot_carbon_material_properties(
    save_path="carbon_material_properties.pdf",
    dpi=600,
):
    """
    Plot the carbon OCV, exchange-current density, diffusivity,
    conductivity, and chemical potential used by BESFEM.
    """

    if graphite_ticks is None or graphite_i0_table is None:
        print(
            "Skipping the carbon property figure because carbon_i0 "
            "currently reuses the graphite exchange-current table, "
            "which was not found."
        )
        return None, None

    x_si = np.linspace(0.0, 1.0, 1000)

    properties = {
        "ocv": carbon_ocv(x_si),
        "i0": carbon_i0(x_si),
        "diffusivity": carbon_diff(x_si),
        "conductivity": carbon_conductivity(x_si),
        "mu": carbon_mu(x_si),
    }

    fig, axes = plt.subplots(
        nrows=3,
        ncols=2,
        figsize=(7.2, 8.0),
        sharex=True,
        constrained_layout=True,
    )

    ax_ocv, ax_i0 = axes[0]
    ax_diff, ax_cond = axes[1]
    ax_mu = axes[2, 0]
    ax_info = axes[2, 1]

    ax_ocv.plot(x_si, properties["ocv"], color="black")
    ax_ocv.set_ylabel(r"Open-circuit voltage, $U_{\mathrm{eq}}$ [V]")
    ax_ocv.set_title("(a) Equilibrium potential", loc="left")

    ax_i0.plot(x_si, properties["i0"], color="black")
    ax_i0.set_ylabel(r"Exchange current density, $i_0$ [A cm$^{-2}$]")
    ax_i0.set_title("(b) Exchange current density", loc="left")

    ax_diff.plot(x_si, properties["diffusivity"], color="black")
    ax_diff.set_ylabel(r"Diffusivity, $D$ [cm$^2$ s$^{-1}$]")
    ax_diff.set_title("(c) Lithium diffusivity", loc="left")

    ax_cond.plot(x_si, properties["conductivity"], color="black")
    ax_cond.set_ylabel(r"Conductivity, $\sigma$ [S cm$^{-1}$]")
    ax_cond.set_title("(d) Electronic conductivity", loc="left")

    ax_mu.plot(x_si, properties["mu"], color="black")
    ax_mu.set_ylabel(r"Chemical potential, $\mu$")
    ax_mu.set_title("(e) Chemical potential", loc="left")

    ax_info.axis("off")
    ax_info.text(
        0.03,
        0.92,
        "Current BESFEM constants",
        transform=ax_info.transAxes,
        va="top",
        fontweight="bold",
    )
    ax_info.text(
        0.03,
        0.76,
        r"$D_{\mathrm{Si}} = 1.0\times10^{-10}$ cm$^2$ s$^{-1}$"
        "\n"
        r"$\sigma_{\mathrm{Si}} = 1.0$ S cm$^{-1}$"
        "\n"
        r"$\rho_{\mathrm{Si}} = 0.0227$"
        "\n\n"
        r"$i_{0,\mathrm{Si}}$ currently reuses $i_{0,\mathrm{graphite}}$."
        "\n"
        "carbon mobility is not currently defined.",
        transform=ax_info.transAxes,
        va="top",
        linespacing=1.5,
    )

    plotted_axes = [ax_ocv, ax_i0, ax_diff, ax_cond, ax_mu]

    for ax in plotted_axes:
        ax.set_xlim(0.0, 1.0)
        ax.tick_params(
            axis="both",
            which="major",
            direction="in",
            top=True,
            right=True,
            length=5,
            width=0.8,
        )
        ax.tick_params(
            axis="both",
            which="minor",
            direction="in",
            top=True,
            right=True,
            length=2.5,
            width=0.6,
        )
        ax.xaxis.set_minor_locator(AutoMinorLocator())
        ax.yaxis.set_minor_locator(AutoMinorLocator())
        ax.grid(True, which="major", linewidth=0.4, alpha=0.25)

    for ax in (ax_i0, ax_diff):
        ax.ticklabel_format(axis="y", style="sci", scilimits=(-2, 2))

    fig.supxlabel(r"Lithium site fraction, $X$")

    fig.savefig(save_path, dpi=dpi, bbox_inches="tight")
    png_path = save_path[:-4] + ".png" if save_path.lower().endswith(".pdf") else save_path + ".png"
    fig.savefig(png_path, dpi=dpi, bbox_inches="tight")

    plt.show()

    print(f"Saved vector figure to: {save_path}")
    print(f"Saved raster figure to: {png_path}")

    return fig, axes


fig_si, axes_si = plot_carbon_material_properties(
    save_path="carbon_material_properties.pdf",
    dpi=600,
)
```

4. electronic conductivity.

The initial blend conditions are marked on the equilibrium-potential and chemical-potential panels:
\(X_{\mathrm{Gr}}=0.20\) for graphite and \(X_{\mathrm{Si}}=0.40\) for carbon.

carbon currently reuses the graphite exchange-current-density table, so those two curves overlap. A direct diffusivity overlay is not included because the present graphite model uses a concentration-dependent mobility for the Cahn--Hilliard equation, whereas carbon uses a diffusivity in a Fickian transport model. These quantities are not directly equivalent.

The chemical-potential panel reproduces the functions currently defined in the notebook. Confirm that the graphite and carbon chemical potentials use consistent units before using the panel for quantitative comparison.


```python
# ============================================================
# GRAPHITE–carbon MATERIAL PROPERTY OVERLAYS
# ============================================================

def plot_graphite_carbon_overlays(
    save_path="graphite_carbon_property_overlays.pdf",
    dpi=600,
    graphite_initial_x=0.20,
    carbon_initial_x=0.40,
):
    """
    Overlay graphite and carbon properties used by BESFEM.

    The initial graphite and carbon lithiation fractions are marked on
    the OCV and chemical-potential panels.
    """

    required_tables = {
        "graphite_ticks": graphite_ticks,
        "graphite_ocv_table": graphite_ocv_table,
        "graphite_i0_table": graphite_i0_table,
        "graphite_mu_table": graphite_mu_table,
    }

    missing_tables = [
        name for name, values in required_tables.items()
        if values is None
    ]

    if missing_tables:
        raise FileNotFoundError(
            "Cannot create the graphite–carbon overlay because the "
            "following graphite tables were not loaded: "
            + ", ".join(missing_tables)
        )

    x = np.linspace(0.0, 1.0, 1000)

    # Evaluate all shared properties.
    graphite_values = {
        "ocv": graphite_ocv(x),
        "mu": graphite_mu(x),
        "i0": graphite_i0(x),
        "conductivity": graphite_conductivity(x),
    }

    carbon_values = {
        "ocv": carbon_ocv(x),
        "mu": carbon_mu(x),
        "i0": carbon_i0(x),
        "conductivity": carbon_conductivity(x),
    }

    # Initial-state values for markers.
    gr_ocv_0 = float(np.asarray(graphite_ocv(graphite_initial_x)))
    si_ocv_0 = float(np.asarray(carbon_ocv(carbon_initial_x)))
    gr_mu_0 = float(np.asarray(graphite_mu(graphite_initial_x)))
    si_mu_0 = float(np.asarray(carbon_mu(carbon_initial_x)))

    plt.rcParams.update({
        "font.family": "serif",
        "font.serif": ["Times New Roman", "Times", "DejaVu Serif"],
        "mathtext.fontset": "stix",
        "font.size": 10,
        "axes.labelsize": 11,
        "axes.titlesize": 11,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 8.5,
        "axes.linewidth": 0.8,
        "lines.linewidth": 1.8,
        "savefig.bbox": "tight",
    })

    fig, axes = plt.subplots(
        nrows=2,
        ncols=2,
        figsize=(7.4, 6.0),
        sharex=True,
        constrained_layout=True,
    )

    ax_ocv, ax_mu = axes[0]
    ax_i0, ax_cond = axes[1]

    # --------------------------------------------------------
    # (a) Equilibrium potential
    # --------------------------------------------------------
    ax_ocv.plot(
        x,
        graphite_values["ocv"],
        color="black",
        linestyle="-",
        label="Graphite",
    )
    ax_ocv.plot(
        x,
        carbon_values["ocv"],
        color="red",
        linestyle="--",
        label="Carbon",
    )
    ax_ocv.scatter(
        graphite_initial_x,
        gr_ocv_0,
        color="black",
        marker="o",
        s=28,
        zorder=5,
        label=rf"Graphite initial, $X={graphite_initial_x:.2f}$",
    )
    ax_ocv.scatter(
        carbon_initial_x,
        si_ocv_0,
        color="red",
        marker="s",
        s=28,
        zorder=5,
        label=rf"Carbon initial, $X={carbon_initial_x:.2f}$",
    )
    ax_ocv.set_ylabel(r"Equilibrium potential, $U_{\mathrm{eq}}$ [V]")
    ax_ocv.set_title("(a) Equilibrium potential", loc="left")

    # --------------------------------------------------------
    # (b) Chemical potential
    # --------------------------------------------------------
    ax_mu.plot(
        x,
        graphite_values["mu"],
        color="black",
        linestyle="-",
        label="Graphite",
    )
    ax_mu.plot(
        x,
        carbon_values["mu"],
        color="red",
        linestyle="--",
        label="Carbon",
    )
    ax_mu.scatter(
        graphite_initial_x,
        gr_mu_0,
        color="black",
        marker="o",
        s=28,
        zorder=5,
    )
    ax_mu.scatter(
        carbon_initial_x,
        si_mu_0,
        color="red",
        marker="s",
        s=28,
        zorder=5,
    )
    ax_mu.set_ylabel(r"Chemical potential, $\mu$")
    ax_mu.set_title("(b) Chemical potential", loc="left")

    # --------------------------------------------------------
    # (c) Exchange-current density
    # --------------------------------------------------------
    ax_i0.plot(
        x,
        graphite_values["i0"],
        color="black",
        linestyle="-",
        linewidth=2.4,
        label="Graphite",
    )
    ax_i0.plot(
        x,
        carbon_values["i0"],
        color="red",
        linestyle="--",
        linewidth=1.5,
        label="Carbon",
    )
    ax_i0.set_ylabel(
        r"Exchange current density, $i_0$ [A cm$^{-2}$]"
    )
    ax_i0.set_title("(c) Exchange-current density", loc="left")
    ax_i0.ticklabel_format(
        axis="y",
        style="sci",
        scilimits=(-2, 2),
    )

    # --------------------------------------------------------
    # (d) Electronic conductivity
    # --------------------------------------------------------
    ax_cond.plot(
        x,
        graphite_values["conductivity"],
        color="black",
        linestyle="-",
        label="Graphite",
    )
    ax_cond.plot(
        x,
        carbon_values["conductivity"],
        color="red",
        linestyle="--",
        label="Carbon",
    )
    ax_cond.set_ylabel(
        r"Electronic conductivity, $\sigma$ [S cm$^{-1}$]"
    )
    ax_cond.set_title("(d) Electronic conductivity", loc="left")

    # --------------------------------------------------------
    # Shared formatting
    # --------------------------------------------------------
    for ax in axes.flat:
        ax.set_xlim(0.0, 1.0)
        ax.tick_params(
            axis="both",
            which="major",
            direction="in",
            top=True,
            right=True,
            length=5,
            width=0.8,
        )
        ax.tick_params(
            axis="both",
            which="minor",
            direction="in",
            top=True,
            right=True,
            length=2.5,
            width=0.6,
        )
        ax.xaxis.set_minor_locator(AutoMinorLocator())
        ax.yaxis.set_minor_locator(AutoMinorLocator())
        ax.grid(
            True,
            which="major",
            linewidth=0.4,
            alpha=0.25,
        )
        ax.legend(frameon=False, loc="best")

    fig.supxlabel(r"Lithium site fraction, $X$")

    # --------------------------------------------------------
    # Save and display
    # --------------------------------------------------------
    fig.savefig(
        save_path,
        dpi=dpi,
        bbox_inches="tight",
    )

    if save_path.lower().endswith(".pdf"):
        png_path = save_path[:-4] + ".png"
    else:
        png_path = save_path + ".png"

    fig.savefig(
        png_path,
        dpi=dpi,
        bbox_inches="tight",
    )

    plt.show()

    print(f"Saved vector figure to: {save_path}")
    print(f"Saved raster figure to: {png_path}")
    print()
    print("Initial-state values:")
    print(
        f"  Graphite: X = {graphite_initial_x:.3f}, "
        f"Ueq = {gr_ocv_0:.6g}, mu = {gr_mu_0:.6g}"
    )
    print(
        f"  carbon:  X = {carbon_initial_x:.3f}, "
        f"Ueq = {si_ocv_0:.6g}, mu = {si_mu_0:.6g}"
    )
    print()
    print(
        "The exchange-current curves overlap because carbon currently "
        "reuses the graphite exchange-current table."
    )

    return fig, axes


fig_gr_si, axes_gr_si = plot_graphite_carbon_overlays(
    save_path="graphite_carbon_property_overlays.pdf",
    dpi=600,
    graphite_initial_x=0.20,
    carbon_initial_x=0.40,
)
```
